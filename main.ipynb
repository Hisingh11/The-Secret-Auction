```python
import os
import random
import import_ipynb
from art import logo

def clear():
    os.system('cls' if os.name == 'nt' else 'clear')

bid = {}
continu_bidding = True
print(logo)

def find_highest_amount(bidding_details):
    highest_bid = 0  
    winner = ""      
    
    for bidder in bidding_details:
        bid_amount = bidding_details[bidder]
        if bid_amount > highest_bid:
            highest_bid = bid_amount
            winner = bidder
            
    
    print(f"The winner is {winner} with the winning bid of ${highest_bid}")

while continu_bidding:
    bidder_name = input("Enter your name: ")
    bid_price = int(input("Enter your bid amount: $"))
    bid[bidder_name] = bid_price
    
    should_continue = input("Are there any new bidders? Type 'yes' or 'no'.\n").lower()
    
    if should_continue == "no":
        continu_bidding = False
        find_highest_amount(bid)  
    else: 
        clear() 
```
